4. У якому місяці мережа отримує найменше сумарного доходу (Total) та валового прибутку (gross income) і які дні тижня просідають?

```python
df['Month'] = pd.to_datetime(df['Date']).dt.month

df.groupby('Month')[['Total', 'gross income']].sum().round(2).sort_values(by='Total')
```

```python
df['Day_of_week'] = pd.to_datetime(df['Date']).dt.day_name()

df.groupby('Day_of_week')['Total'].agg(
    Check_count='count',
    Mean_check='mean',
    Total_revenue='sum'
).round(2).sort_values(by='Total_revenue')
```

Найменш прибутковий місяць: лютий з найнижчими показниками загального доходу та  валового прибутку. Найменш прибутковий день: понеділок, який суттєво просідає, як за кількістю чеків, так і за середнім чеком, формуючи найнижчий загальний дохід.

Використання результатів для прийняття бізнес-рішень:
- стимулювати прибуток у дні з найнижчими показниками (понеділок): запустити акцію «Щасливий понеділок» (наприклад, подвійні бали в програмі лояльності або спеціальна знижка на популярні категорії товарів);
- направляти таргетовані push-сповіщення у неділю ввечері із закликом зробити закупки на початок тижня;
- підвищити середній чек в понеділок та середу (де середній чек менший за 306 дол.) впровадженням акції за типом "купи на суму від 350 дол. та отримай подарунок/знижку".
- заздалегідь планувати маркетинговий календар на лютий (сезонні розпродажі, тематичні промо-кампанії до свят), щоб вирівняти просідання виручки порівняно із січнем та березнем.

5. Як тип клієнта (Customer type) впливає на розмір середнього чека у різних категоріях товарів (Product line)?

```python
customer_category_analysis = pd.pivot_table(
    df,
    index='Product line',
    columns='Customer type',
    values='Total',
    aggfunc='mean'
).round(2)

customer_category_analysis['Difference (Member - Normal)'] = (
    customer_category_analysis['Member'] - customer_category_analysis['Normal']
).round(2)

customer_category_analysis
```

Використання результатів для прийняття бізнес-рішень:

- якщо у певній категорії середній чек звичайних покупців (Normal) вищий, ніж у учасників програми лояльності (Member), це сигнал, що клієнт не бачать цінності купувати дорогі товари з цієї категорії саме в нашій мережі. Потрібно ввести спеціальні бонуси/кешбек на ці категорії для Member(згідно таблиці категорії Electronic accessories та Sports and travel мають середній чек у клієнтів Normal вищий);

- для категорій, де Member демонструють значно вищий середній чек, варто запускати ексклюзивні попередні продажі (Early Access) чи персналізовані e-mail/push-розсилки з новинками(наприклад: в категорії Health and beauty);

- на касі можна пропонувати оформлення картки лояльності з акцентом на ті категорії, де члени клубу купують значно більше. Наприклад: «Зазвичай наші постійні клієнти платять менше за товари з категорії Health and beauty завдяки персональним бонусам — бажаєте оформити картку та заощадити вже зараз?»

6. Як дохід та середній чек відрізняється між клієнтами, а саме чоловіками (Male) та жінками (Female) залежно від категорії товарів (Product line)?

```python
gender_product = pd.pivot_table(
    df,
    index='Product line',
    columns='Gender',
    values='Total',
    aggfunc=['sum', 'mean']
).round(2)

gender_product
```
